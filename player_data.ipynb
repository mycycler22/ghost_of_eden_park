```python
from rugbypy.match import *
from rugbypy.competition import *
from rugbypy.team import *
from rugbypy.venue import *
from rugbypy.player import *
```

```python
import pandas as pd
import polars as pl
import matplotlib.pyplot as plt
import numpy as np
```

```python
test_nations = pd.read_excel("test_nations.xlsx")

test_nations
```

```python
all_team_stats = pd.DataFrame()
team_ids = test_nations['team_id']

for id in team_ids:
    all_team_stats = pd.concat([all_team_stats, fetch_team_stats(team_id=id)], ignore_index=True)
```

```python
all_team_stats
```

```python
all_players = all_team_stats[['players']]

all_players
```

```python
all_players = all_players[all_players['players'].str.len() > 1]


all_players
```

```python
type(all_players)
```

```python
from itertools import chain

player_stats = pd.DataFrame()

for id in chain.from_iterable(all_players['players']):
    player_stats = pd.concat([player_stats, fetch_player_stats(player_id=id)])
```

```python
test_player_stats = player_stats[player_stats['team_id'].isin(test_nations['team_id'])]


test_player_stats
```
