```python
from rugbypy.match import *
from rugbypy.competition import *
from rugbypy.team import *
from rugbypy.venue import *
from rugbypy.player import *
```

```python
import pandas as pd
import polars as pl
import matplotlib.pyplot as plt
import numpy as np
```

```python
test_nations = pd.read_excel("test_nations.xlsx")

test_nations
```

```python
all_team_stats = pd.DataFrame()
team_ids = test_nations['team_id']

for id in team_ids:
    all_team_stats = pd.concat([all_team_stats, fetch_team_stats(team_id=id)], ignore_index=True)
```

```python
all_team_stats
```

```python
all_players = all_team_stats[['game_date', 'players']]

all_players
```

```python
all_players = all_players[all_players['players'].str.len() > 1]


all_players
```

```python
player_stats = pd.DataFrame()
dfs = []

for date, ids in zip(all_players['game_date'], all_players['players']):
    for id in ids:
        dfs.append(
            fetch_player_stats(
                player_id = id,
                date = date
            )
        )

player_stats = pd.concat(dfs, ignore_index=True)
```

```python
player_stats
```

```python
## Only including test matches

match_stats = player_stats[player_stats['team_id'].isin(test_nations['team_id'])]

match_stats
```

```python
## Aggregating stats to collect team totals

match_stats = (match_stats.groupby(["team", "match_id"], as_index=False)
      .sum(numeric_only=True)
)

match_stats.head()
```

```python
match_stats.to_excel("aggregated_stats.xlsx", index=False)
```
